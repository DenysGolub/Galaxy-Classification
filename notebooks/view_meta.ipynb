```python
metadata_df = pd.read_csv("../data/metadata_new.csv")

full_df = pd.DataFrame()
decals_df = pd.DataFrame({
    "ra": ra,
    "dec": dec,
    "images": list(images),
    "redshift": redshift,
    "pxscale": pxscale,
    "galaxy_type": labels
})
```

```python
metadata_df["ra_round"] = metadata_df["ra"].round(5)
metadata_df["dec_round"] = metadata_df["dec"].round(5)

decals_df["ra_round"] = decals_df["ra"].round(5)
decals_df["dec_round"] = decals_df["dec"].round(5)

full_df = pd.merge(
    decals_df,
    metadata_df,
    on=["ra_round", "dec_round"],
    how="left"
)
```

```python
full_df.dropna(inplace=True)
```

```python
metadata_df = metadata_df.drop_duplicates(subset=["ra_round", "dec_round"])
```

```python
full_df.drop(['ra_y', 'dec_y', "ra_round", "dec_round"], axis=1, inplace=True)
```

```python
full_df.rename(columns={"ra_x": "ra", "dec_x": "dec"}, inplace=True)
```

```python
full_df.head()
```

```python

# Original labels are 0-9, we map them to 0-3 based on the provided mapping.
# 0,1 -> 0 (Disturbed / Merging)
# 2,3,4 -> 1 (Smooth)
# 5,6,7 -> 2 (Spiral)
# 8,9 -> 3 (Edge-on)

full_df['galaxy_type'] = full_df['galaxy_type'].map({0: 0, 
                                                     1: 0, 
                                                     2: 1, 
                                                     3: 1, 
                                                     4: 1, 
                                                     5: 2, 
                                                     6: 2, 
                                                     7: 2, 
                                                     8: 3, 
                                                     9: 3})
```

```python
full_df.value_counts('galaxy_type')
```

```python
import h5py
import numpy as np

image_tensor = np.stack(full_df['images'].values).astype('float32')

with h5py.File('../data/galaxies_multimodal.h5', 'w') as hf:
    
    hf.create_dataset('images', data=image_tensor, compression="gzip", chunks=True)
    
    for col in full_df.columns:
        if col == 'images':
            continue  
            
        data = full_df[col].values
        
        if full_df[col].dtype == 'object' or isinstance(data[0], str):
            # Convert to S-type (fixed-length byte strings)
            clean_data = [str(x).encode('utf-8') for x in data]
            hf.create_dataset(col, data=clean_data)
            
        else:
            clean_data = full_df[col].fillna(-999).values
            hf.create_dataset(col, data=clean_data)

print("HDF5 file created successfully with all columns.")
```
